```python
#write a python program to find number of duplicates in a list
l1=list(input().split())
s1=set(l1)
print(len(l1)-len(s1))
```

```python
#write a program to find no.of elements having duplicates
l2=list(input().split())
s2=set(l2)
l3=list(s2)
print(l2)
c=0
for i in range(len(l3)):
   if(l2.count(l3[i])>1):
    c=c+1
print(c)
```

```python
#write a python program that prints missing between 1 to n
n=int(input())
list1=list(map(int,input().split()))
for i in range(1,n+1):
  if(i not in list1):
    print(i,end=" ")
```
